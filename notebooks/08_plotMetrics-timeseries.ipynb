```python
import xarray as xr
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings('ignore')

import matplotlib.cm as cm
# fontsize = 40
# plt.rcParams.update({'font.size': fontsize}) 
from cmap import Colormap
import seaborn as sns
from scipy.stats import pearsonr
```

```python
path = "/pub/mariant3/WarmWaterMassesv1/"
metrics = xr.open_dataset(f"{path}data/mhwMetrics.nc")
```

```python
fontsize = 30

years = ['0186', '0187', '0188', '0189']

fig, axs = plt.subplots(len(years), 2, figsize=(30, 35))

color_sets = {
    '0186': sns.color_palette("Dark2", n_colors=8),
    '0187': sns.color_palette("husl", n_colors=8),
    '0188': sns.color_palette("Set1", n_colors=4),
    '0189': sns.color_palette("Set2", n_colors=8)
}

# linestyles = [
#     "-",      
#     "--", 
#     "-.",          
#     ":",                  
#     (0, (10, 3)),        
#     (0, (8, 2, 2, 2)),  
#     (0, (14, 4)),       
#     (0, (12, 2, 2, 2)),   
# ]
linestyles = [
    "-",                  
    "-",                  
   "-",                  
   "-",                  
   "-",                  
   "-",                 
   "-",                  
   "-",                
]

markers = [
    "o",
    "s",  
    "^", 
    "D",
    "P",  
    "X",  
    "v",  
    "*",  
]

for i, year in enumerate(years):

    year_big_events = (
        metrics.sel(time=year)
        .event_area
        .where(metrics.sel(time=year).event_area > 0.15e12, drop=True)
        .mhw
    )

    for j, mhw in enumerate(year_big_events):

        color = color_sets[year][j % len(color_sets[year])]
        linestyle = linestyles[j % len(linestyles)]
        marker = markers[j % len(markers)]

        big_areas = metrics.sel(mhw=mhw, time=year).event_area
        big_volumes = metrics.sel(mhw=mhw, time=year).event_volume

        big_areas.plot(
            ax=axs[i,0],
            color=color,
            linestyle=linestyle,
            linewidth=8,
            marker=marker,
            markersize=18,
            markevery=16,
            markeredgecolor='k',
            markeredgewidth=1.2,
            alpha=1,
            label=f'Event {mhw.values}'
        )

        big_volumes.plot(
            ax=axs[i,1],
            color=color,
            linestyle=linestyle,
            linewidth=8,
            marker=marker,
            markersize=18,
            markevery=16,
            markeredgecolor='k',
            markeredgewidth=1.2,
            alpha=1
        )

    for j in range(2):

        axs[i,j].set_xlim(
            cftime.DatetimeNoLeap(int(year), 3, 22),
            cftime.DatetimeNoLeap(int(year), 11, 15)
        )

        axs[i,j].tick_params(axis='both', labelsize=30)
        axs[i,j].grid(True, alpha=0.3)
        axs[i,j].set_xlabel("")
        axs[i,j].set_ylabel("")
        axs[i,j].set_title("")

    axs[i,0].set_yticks(np.arange(0, 6e12, 1e12))
    axs[i,0].set_ylim(0, 3e12)

    axs[i,1].set_yticks(np.arange(0, 8e13, 1e13))
    axs[i,1].set_ylim(0, 5e13)

    axs[i,0].yaxis.get_offset_text().set_color('w')
    axs[i,1].yaxis.get_offset_text().set_color('w')

    axs[i,0].legend(
        loc='upper left',
        fontsize=25,
        framealpha=0.9,
        ncol=1
    )

    axs[i,0].text(
        1, 0.98,
        f"(a.{i+1})",
        transform=axs[i,0].transAxes,
        fontsize=25,
        va="top",
        ha="right"
    )

    axs[i,1].text(
        1, 0.98,
        f"(b.{i+1})",
        transform=axs[i,1].transAxes,
        fontsize=25,
        va="top",
        ha="right"
    )

fig.text(
    0.04,
    0.5,
    r"Area [10$^{12}$ m$^{2}$]",
    va='center',
    rotation='vertical',
    fontsize=30
)

fig.text(
    0.51,
    0.5,
    r"Volume [10$^{13}$ m$^{3}$]",
    va='center',
    rotation='vertical',
    fontsize=30
)

plt.tight_layout(rect=[0.03, 0, 1, 1], pad=3)

# plt.savefig(
#     "../figures/area-volume-timeseries-25-accessible_1.png",
#     dpi=400,
#     bbox_inches="tight"
# )

plt.show()
```

```python
fontsize = 30

years = ['0186', '0187', '0188', '0189']

fig, axs = plt.subplots(len(years), 2, figsize=(30, 35))

color_sets = {
    '0186': sns.color_palette("Dark2", n_colors=8),
    '0187': sns.color_palette("husl", n_colors=8),
    '0188': sns.color_palette("Set1", n_colors=4),
    '0189': sns.color_palette("Set2", n_colors=8)
}

markers = [
    "o",
    "s",
    "^",
    "D",
    "P",
    "X",
    "v",
    "*",
]

for i, year in enumerate(years):

    year_big_events = (
        metrics.sel(time=year)
        .event_area
        .where(metrics.sel(time=year).event_area > 0.15e12, drop=True)
        .mhw
    )

    for j, mhw in enumerate(year_big_events):

        color = color_sets[year][j % len(color_sets[year])]
        marker = markers[j % len(markers)]

        big_thickness = metrics.sel(
            mhw=mhw,
            time=year
        ).event_thickness

        big_T = metrics.sel(
            mhw=mhw,
            time=year
        ).volcello_weighted_thetao

        big_thickness.plot(
            ax=axs[i,0],
            color=color,
            linewidth=8,
            marker=marker,
            markersize=18,
            markevery=16,
            markeredgecolor='k',
            markeredgewidth=1.2,
            alpha=1,
            label=f'Event {mhw.values}'
        )

        big_T.plot(
            ax=axs[i,1],
            color=color,
            linewidth=8,
            marker=marker,
            markersize=18,
            markevery=16,
            markeredgecolor='k',
            markeredgewidth=1.2,
            alpha=1
        )

    for j in range(2):

        axs[i,j].set_xlim(
            cftime.DatetimeNoLeap(int(year), 3, 22),
            cftime.DatetimeNoLeap(int(year), 11, 15)
        )

        axs[i,j].tick_params(
            axis='both',
            labelsize=30
        )

        axs[i,j].grid(True, alpha=0.3)

        axs[i,j].set_xlabel("")
        axs[i,j].set_ylabel("")
        axs[i,j].set_title("")

    # Thickness
    axs[i,0].set_yticks(np.arange(0, 31, 5))
    axs[i,0].set_ylim(0, 30)
    axs[i,0].invert_yaxis()

    # Volume-weighted temperature
    axs[i,1].set_yticks(np.arange(29, 31.1, 0.5))
    axs[i,1].set_ylim(29, 31)

    axs[i,0].legend(
        loc='lower left',
        fontsize=25,
        framealpha=0.5,
        ncol=1
    )

    axs[i,0].text(
        1,
        0.98,
        f"(c.{i+1})",
        transform=axs[i,0].transAxes,
        fontsize=25,
        va="top",
        ha="right"
    )

    axs[i,1].text(
        1,
        0.98,
        f"(d.{i+1})",
        transform=axs[i,1].transAxes,
        fontsize=1,
        va="top",
        ha="right"
    )

fig.text(
    0.04,
    0.5,
    "Thickness [m]",
    va='center',
    rotation='vertical',
    fontsize=30
)

fig.text(
    0.51,
    0.5,
    r"Volume averaged temperature [$^\circ$C]",
    va='center',
    rotation='vertical',
    fontsize=30
)

plt.tight_layout(
    rect=[0.03, 0, 1, 1],
    pad=3
)

# plt.savefig(
#     "../figures/weightedTemp-thickness-timeseries-accessible_1.png",
#     dpi=400,
#     bbox_inches="tight"
# )

plt.show()
```

```python
df = pd.DataFrame({
    'area': metrics.event_area.max(dim='time').values,
    'volume': metrics.event_volume.max(dim='time').values
})

r, p_value = pearsonr(df['area'], df['volume'])
r_squared = r**2

ax = sns.regplot(data=df, x='area', y='volume', scatter_kws={'alpha':1}, 
                 label=f'R² = {r_squared:.2f}')

plt.xlabel(r'Area [10$^{12}$ m²]')
plt.ylabel(r'Volume [10$^{13}$ m$^{3}]$')
plt.legend(loc='lower right', fontsize=15)
plt.show()
```
